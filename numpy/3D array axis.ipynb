```python
np_array_2 = np.arange(27,54).reshape(3,3,3)
np_array_2
```

### When axis is 0
- The sum of inner array will happen element wise for example 
        [[ 0,  1,  2],        [[ 9, 10, 11],      [[18, 19, 20],
        [ 3,  4,  5],    +    [12, 13, 14],   +   [21, 22, 23],
        [ 6,  7,  8]]         [15, 16, 17]]       [24, 25, 26]]
    The three inner arrays will be summed element wise like 
    - 1+9+18 = 27
    - 1+10+19 = 30

```python
np_array_1
```

```python
axis_0_sum = np.sum(np_array_1,axis=0)
axis_0_sum
```

### When axis is 1
- The sum will happen column wise in each inner matrix and the result of that sum will be saved in the resultant array

```python
np_array_1
```

```python
axis_1_sum = np.sum(np_array_1,axis=1)
axis_1_sum
```

### When axis is 2
- The sum will happen row wise in each inner matrix and the result of that sum will be saved in the resultant array

```python
np_array_1
```

```python
axis_1_sum = np.sum(np_array_1,axis=2)
axis_1_sum
```
