```python
import numpy as np
```

```python
a = np.arange(12).reshape(3,4)
a
```

## The simple way to iterate through the array is using for loop:

### This will print all the rows in the array

```python
for row in a:
    print(row)
```

### This will print all the elements in the array

```python
for row in a:
    for cell in row:
        print(cell)
```
