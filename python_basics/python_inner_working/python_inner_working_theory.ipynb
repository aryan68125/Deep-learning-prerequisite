  - If nothing in the source file has changed, Python reuses the .pyc file.

### Difference between Java byte code and Python byte code 
| Feature               | Java Bytecode (`.class`)                | Python Bytecode (`.pyc`)                        |
| --------------------- | --------------------------------------- | ----------------------------------------------- |
| 🔄 Compiled From      | `.java` source files                    | `.py` source files                              |
| 📁 Output File        | `.class` files                          | `.pyc` files (in `__pycache__/`)                |
| 💻 Execution Platform | Java Virtual Machine (JVM)              | Python Virtual Machine (PVM)                    |
| 🌐 Portability        | Very high – runs on any JVM             | Limited – version-dependent on Python           |
| ⚙️ Compilation Step   | Explicit (`javac MyClass.java`)         | Implicit (on first import or run of `.py` file) |
| 🧠 Optimization Level | Can be optimized (`-O`) with JIT in JVM | Basic – no runtime optimization                 |
| 🔁 Reusability        | Reusable across platforms via JVM       | Tied to specific Python version (e.g. CPython)  |
| 📦 Packaging          | Can be bundled into `.jar` files        | Not usually bundled – regenerated on demand     |



### NOTE:
- Python virtual machine is the one that actually runs your code under the hood.
- Just because python is generating byte code doesn't mean it is a compiled time language no! Python is an interpreted language.
- Python byte code .pyc file runs faster than the python program because all the checks related to syntax,  Parsing etc. are mostly done.
- Python uses a diffing algorithm to tell if there are any changes made in the script based on what is present in the .pyc file generated from that script. Then python puts the changes made in the script in that .pyc file. Whole python byte code is not generated only new changes are put inside of it.
- When generating a .pyc Python bytecode file, the Python version plays an important role because the .pyc file is specific to the version of Python that created it, and it may not be compatible with other versions during execution.
- These .pyc files are only visible when you import something
- .pyc file (Python Byte code) is not a machine code. They are not the insttuctions that can be fed directly fed in your intel or apple arm processors.
- .pyc is Python sepcific interpretation. 

## Python Virtual Machine PVM
The Python Virtual Machine (PVM) is the runtime engine that interprets and executes Python bytecode (.pyc files). It’s a part of the Python interpreter and plays a central role in Python's execution model.
#### Working of PVM:
- You write Python code (.py file).
- The Python compiler converts that code into bytecode (.pyc file), an intermediate, low-level representation.
- The PVM reads this bytecode and executes it line by line.
- Since PVM executes the .pyc file line by line hence python is called an interpreted language
- PVM can accept .pyc file or directly .py file which is essentially your python script. That's why .pyc file is doesn't get generated for the top level file i.e if a folder has only one python file .pyc file will not get generated.

#### Stale Import in Jupyter (Caching)
When you import a module in Jupyter, it gets cached in memory. If you edit the .py file afterward, those changes don’t reflect unless you explicitly reload the module. <br>
So even though your file now has coffee_one, the version already loaded in Jupyter doesn’t.

`````python
import python_inner_working
import importlib
# Reload the module to apply latest file changes
importlib.reload(python_inner_working)

# Now it should work
python_inner_working.coffee_one
`````

#### As you can see not the code works like a charm

### Explaination:

When I am working in Jupyter Notebook (or any long-running Python session), modules are cached the first time you import them. If I later edit the module's .py file outside the notebook (like in VS Code or a text editor), those changes won’t automatically show up when I am re-import it. <br>
#### Why ```import python_inner_working``` alone is not enough
After editing the file ```python_inner_working.py```, Python will use the old cached version. It won’t re-read the file from disk.
#### ```importlib.reload()```: The Fix
To force python to reload the updated file from the disk use : 
```python
    import importlib
    importlib.reload(python_inner_working)
```
