### Calculate standard deviation
- Standard Deviation (σ) is a statistical measure that tells you how spread out the values in a dataset are from the mean (average)
- If the values are close to the mean, the standard deviation is small.
- If the values are spread out, the standard deviation is large..

### Why standard deviation is important
- Detecting anomalies and outliers:
  - If a data point is several standard deviation away from the mean, it's likely and outlier.
  - It is used for Fraud detection, quality control, sensor anomalies.
- Comparing performance consistency:
  - Let's say I have to analyze the test score of two students over several exams:
  - Student A has more consistent performance.
  - Student B has higher variability.
  - Useful for: HR appraisals, academic evaluations, product quality
- Risk assesment in Finance:
  - Standar deviation is often used as a measure of risk in investment returns.
  - Higher standard deviation means more unpredictable returns and higher risks.
  - Useful for portfolio management, stock analysis.
- Understanding spread in distributions:
  - You can visualize it using histogram + standard ranges to understand how data clusters around the mean

### Ok so lets create a function in python that calculates the bin automatically by getting the count of the data from the desired column from the dataFrame and apply the Sturges' Rule to find the optimum value for bin to create a histogram

````python
import math
def calculate_bin_value_using_sturge(data_point_count):
    if data_point_count<=0:
        raise ValueError("Number of data points must be grater than zero")
    # math.ceil() ensures you round up to the nearest whole number, 
    # ensuring the number of bins is enough to cover the data distribution properly.
    return math.ceil(math.log2(data_point_count)+1)
````

#### Why are we doing math.ceil() ?
According to what I read in the pdf sturge's rule equation to find the number of bins is $k = 1 + log_2(n)$ where ```n``` is the number of data points and ```k``` is the bin number. So I can just implement this equation like this math.log2(data_point_count) and be done with it. 
<br>
The reason we are doing math.ceil() is because the result of math.log2() will be in floating point number and bins can't be in float the bin number must be discreet. There are two ways to achieve a discreet number from a floating point number 
- using math.floor() : The reason for not using this is simply because math.floor() round down the float value. which might compress the data too much and hide details.
- using math.cleil() : The reason for using this is simply because math.ceil() round up the float value hence every data point is accomodated because this never underestimate the number of bins.

````python
calculate_bin_value_using_sturge(31)
````

#### Now let's see how the histogram will look if I use Freedman and Diaconis’s (1981) rule
The formula of Freedman and Diaconis’s (1981) rule looks something like this: <br>

#### Interpretation of this formula
### $h = 2.IQR.n^{\frac {-1}{3}}$ <br>
- h = bin width
- IQR = Interquartile range (Q3 - Q1), a robust measure of variability
- n = number of data points
- $n^{\frac {-1}{3}}$ scales the bin width relative to data size

#### What is Interquartile Range (IQR)
The Interquartile Range (IQR) is a measure of statistical dispersion, or how spread out the data is. It specifically focuses on the middle 50% of your data. <br>
IQR=Q3−Q1 <br>
- Q1 (1st Quartile): 25% of the data lies below this value. (in pandas Q1 is calculated by .quantile(0.25))
- Q3 (3rd Quartile): 75% of the data lies below this value. (in pandas Q3 is calculated by .quantile(0.75))
- The difference (Q3 - Q1) gives you the range of the central 50% of your data

#### Visualization of quartile
Imagine a sorted dataset split into 4 equal parts (quartiles): <br>
```python
|----25%----|----25%----|----25%----|----25%----|
0%         Q1         Q2         Q3         100%
           (25%)     (Median)   (75%)
```
           
- Q1 is the value at the 25th percentile.
- Q2 is the value at the 75th percentile.
- IQR is the box width in a box plot.

````python
# This function calculates the bin value using Freedman and Diaconis’s (1981) rule when the data is positively skewed
# Here Q1 is 25% quantile,  Q3 is 75% quantile and n is the count of data points
def Freedman_and_Diaconis_rule(Q1 = None, Q3 = None, n = None):
    IQR = Q3 - Q1
    scale = pow(n,-(1/3))
    bin_width = 2 * IQR * scale
    return bin_width
````

````python
# Q1 25% quantile 
Q1 = df["SkewedDataPositive"].quantile(0.25)
# Q3 75% quantile 
Q3 = df["SkewedDataPositive"].quantile(0.75)
n = df["SkewedDataPositive"].count()
bin_width = Freedman_and_Diaconis_rule(Q1=Q1,Q3=Q3,n=n)
bin_width
````

##### So that means I need to add this logic to Freedman_and_Diaconis_rule function 

````python
def Freedman_and_Diaconis_rule(Q1 = None, Q3 = None, n = None, max_from_data = None , min_from_data = None):
    if Q1 is None:
        raise ValueError("Q1 quantile not found!")

    if Q3 is None:
        raise ValueError("Q3 quantile not found!")

    if n is None:
        raise ValueError("Total data count not found!")

    if max_from_data is None:
        raise ValueError("maximum value from the dataset not found!")

    if min_from_data is None:
        raise ValueError("minimum value from the dataset not found!")
        
    IQR = Q3 - Q1
    scale = pow(n,-(1/3))
    bin_width = 2 * IQR * scale
    data_range = max_from_data - min_from_data
    number_of_bins = math.ceil(data_range/bin_width)
    return number_of_bins
````

````python
# Q1 25% quantile 
Q1 = df["SkewedDataPositive"].quantile(0.25)
# Q3 75% quantile 
Q3 = df["SkewedDataPositive"].quantile(0.75)
n = df["SkewedDataPositive"].count()
max_from_data = df["SkewedDataPositive"].max()
min_from_data = df["SkewedDataPositive"].min()
bin_number = Freedman_and_Diaconis_rule(Q1=Q1,Q3=Q3,n=n,max_from_data=max_from_data,min_from_data=min_from_data)
bin_number
````

Suppose data_positive = [1.2, 2.5, 0.8, 5.0, 0.4, 7.3, 0.6]
- Step 1 : -data_positive

- Step 2 : +data_positive.max()
  - Get the maximum value of the original data : data_positive.max() = 7.3
  - Add this max value to each of the negated elements : [-1.2 + 7.3, -2.5 + 7.3, ..., -7.3 + 7.3] = [6.1, 4.8, 6.5, 2.3, 6.9, 0.0, 6.7]
- So now:
  - The largest original value (7.3) becomes the smallest (0.0)
  - The smallest values becomes large
  - and the long right tail becomes a long left tail --> Negatively skewed!
