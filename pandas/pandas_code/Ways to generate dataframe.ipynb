## From Python dictionary
 - Create a dataframe using a python dictionary

````python
weather_data = {
    "day":["1/1/2017","1/2/2017","1/3/2017","1/4/2017","1/5/2017","1/6/2017"],
    "temperature":[32,35,28,24,32,31],
    "windspeed":[6,7,2,7,4,2],
    "event":["Rain","Sunny","Snow","Snow","Rain","Sunny"]
}
df_from_dict = pd.DataFrame(weather_data)
df_from_dict
````

## From list of tuples

````python
# convert the weather_data into a list of tuples 
list_of_tuples = list(zip(
    weather_data["day"],
    weather_data["temperature"],
    weather_data["windspeed"],
    weather_data["event"]
))
list_of_tuples
````

````python
# Generating dataframe from the list of tuples
df_from_list_of_tuples = pd.DataFrame(list_of_tuples, columns = ["day","temperature","windspeed","event"])
df_from_list_of_tuples
````

## From the list of dictionaries

````python
# conver the weather data into a list of dictionaries
list_of_dicts = [
    {"day":d,"temperature":t,"windspeed":w,"event":e}
    for d,t,w,e in zip(
        weather_data["day"],
        weather_data["temperature"],
        weather_data["windspeed"],
        weather_data["event"]
    )
]
list_of_dicts
````

````python
# Generate dataframe from the list of dictioanries
df_from_list_of_dict = pd.DataFrame(list_of_dicts,columns = ["day","temperature","windspeed","event"])
df_from_list_of_dict
````

## Working with HDFStore
- HDF5 is a binary file format designed to store large amounts of data efficiently.
- It can store data in a hierarchical structure (like a file system).
### PyTables
- PyTables is a Python library built on top of HDF5.
- It allows querying and manipulating HDF5 files.
### Use Case:
- Storing large datasets with fast read/write access.
- Splitting datasets into named sections (like "train", "test").
- Efficient storage for time series or multi-indexed data.

### Error I got : NotImplementedError: Cannot store a category dtype in a HDF5 dataset that uses format="fixed". Use format="table".
- You're trying to save a category column to HDF5 using format="fixed", which is the default format in df.to_hdf().
- Pandas only supports category dtype in format="table", not in the default format="fixed".

#### What is formats in pandas?
- In Pandas the to_hdf() method supports two HDF5 storage formats "fixed" and "table"
- Fixed format:
  - Default format (if you don't specify any format)
  - very fast for writing and reading
  - Stores DataFrame as a single block --> like a snapshot.
  - Does not support filtering queries
  - Does not support category dtypes
- Table format:
  - Slightly slower to write, but more flexible
  - Stores data in a structured way that supports:
    - Querying rows based on column conditions (like SQL)
    - Appending rows to the file later
    - Storing advanced dtypes like category
    - required if the requirement is to store category columns
    - allows partial loading/filtering
- Use case: When the requirment is to:
  - Filter data using query()
  - append new rows to existing HDF5 data, or store category , object or complex types.

````python
df.to_hdf("weather.h5", key="newyork_weather_data", mode="w", format="table")
````

### Conclusion 

### Why this did not work
- ``` python
    df["PrecipitationIn"] = df["PrecipitationIn"].astype("string")
    df["Events"] = df["Events"].astype("string")
  ```
- This failed when saving to HDF5 (with format="fixed") because:
  - "strings" in pandas creates a StringArray extension type, not a regular Numpy array of strings
  - HDF5 with format="fixed" only supports Numpy native types (int, float, str, etc.) not pandas extension types like StringArray, Categorical, NullableInteger etc.
  - So even though the "string" looks like a string, It is an object with internal structure and HDF5 can't serialize it in "fixed" format.
- That's why the error was "TypeError: objects of type 'StringArray' are not supported in this context"
### Why this did work:
- ```python
    df["PrecipitationIn"] = pd.to_numeric(df["PrecipitationIn"], errors="coerce")
    df["Events"] = df["Events"].astype("category")
  ```
- This worked because:
  - PrecipitationIn as float:
    - converted claenly into float64, which is a native Numpy type that HDF5 supports easily.
  - Events as category:
    - I switched to format = "table" and category is only supported in format="table" (not in the default "fixed" format).
    - Categorical data is compact and effietient and is meant for repeat string values (like weather events).
