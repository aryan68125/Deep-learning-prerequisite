### Error I got : NotImplementedError: Cannot store a category dtype in a HDF5 dataset that uses format="fixed". Use format="table".
- You're trying to save a category column to HDF5 using format="fixed", which is the default format in df.to_hdf().
- Pandas only supports category dtype in format="table", not in the default format="fixed".

#### What is formats in pandas?
- In Pandas the to_hdf() method supports two HDF5 storage formats "fixed" and "table"
- Fixed format:
  - Default format (if you don't specify any format)
  - very fast for writing and reading
  - Stores DataFrame as a single block --> like a snapshot.
  - Does not support filtering queries
  - Does not support category dtypes
- Table format:
  - Slightly slower to write, but more flexible
  - Stores data in a structured way that supports:
    - Querying rows based on column conditions (like SQL)
    - Appending rows to the file later
    - Storing advanced dtypes like category
    - required if the requirement is to store category columns
    - allows partial loading/filtering
- Use case: When the requirment is to:
  - Filter data using query()
  - append new rows to existing HDF5 data, or store category , object or complex types.

### How to deal with feather in dataFrame 
### What is feather?
- Feather is a lightweight, fast, binary columnar file format designed for efficient storage and transfer of large dataframes. It is based on Apache Arrow, which provides a standard in-memory columnar format.
### When is Feather Useful?
- Saving intermediate data during analysis or ML pipelines:
  - Feather is perfect for temporarily saving a DataFrame between data preprocessing and model training.
- Large datasets with performance needs:
  - When you are working with millions of rows csv becomes slow. Feather's binary reads/writes 10-100x faster.
- Multi-Language projects
  - Feather files can be shared between:
    - Python (Pandas)
    - R (data.table / arrow)
    - Julia/C++/Rust via Arrow
  - This makes Feather an excellent choice for cross-language ML or analytics pipelines.
### Some benefits of using feather
- It's much faster
- Less error-prone (no quoting or delimiter issues)
- Better for storage and transfer

### What is Parquet best for and how to use it?
#### 1. Large datasets:
- When your CSV or Excel files becomes slow to read or write or start exceeding memory limits
- Paraquet stores data in columnar format, which means:
  - You only read the columns you need.
  - It's faster and more memory-efficient
#### 2. Repeated use of data
  - Saving intermediate data in pipelines (ETL, ML preprocessing) here ETL means Extrat transform load.
  - Store once --> reuse many times without recomputing.
#### 3. Compressed Storage
  - Parquet supports built-in compression (like snappy, gzip) which:
    - Saves disk space
    - Speeds up loading speed (less I/O)
#### 4. Working with big data tools
  - Parquet is native to:
    - Apache spark
    - Dask
    - Polars
    - Hive, Presto, Athena, Snowflake
#### 5. Column filtering and analytics
  - when you need just a few columns out of many (common in analytics dashboards or api serivces)
  - Columnar structure makes it lightning fast.
#### What is a columnar structure?
  - In a columnar structure data is stored in column by column instead of row by row.
  - +--------+-----+--------+
    | Name   | Age | City   |
    +--------+-----+--------+
    | Alice  | 24  | NYC    |
    | Bob    | 30  | Chicago|
    | Charlie| 28  | Boston |
  - Name:    [Alice, Bob, Charlie]
    Age:     [24, 30, 28]
    City:    [NYC, Chicago, Boston]
  - Lets say I have a datset of 100 million rows and I need to calculate the average salary.
    - In row-based format (CSV)
      - All the columns are read into memory even if you only want the "salary".
    - In columnar format (Parquet):
      - Only the "salary" column is read --- making it faster and more memory efficient.

### ORC behind the scenes

- Very efficient for big data processing.

### What is schema evolution?
- Schema Evolution means the ability to change the structure of data (schema) over time without breaking the existing pipelines.
- So according to what I found if there is a dataFrame with 3 columns in an ORC or paraquet file. Later my app evolves and starts writing more columns or removes or changes existing ones.
- With schema evolution, your system can still read and write data seamlessly, even if the schema has changed across versions of the dataset.

#### Example : (Schema Evolution with paraquet)
#### Version 1 of the data : 

````python
df_v1 = pd.DataFrame({
    "id":[1,2],
    "name":["Alice","Bob"]
})
df_v1.to_parquet("data_v1.parquet")
````

#### New version adds a column

````python
df_v2 = pd.DataFrame({
    "id":[3],
    "name":["charlie"],
    "age":[30]
})
df_v2.to_parquet("data_v2.parquet",append=True)
````

### Analysis as to why I got this error?
- Unlike to_csv() or to_hdf(), the to_paraquet() method does not support appending to an existing Parquet file out of the box.
- Internally to_parquet() calls Parquet engine like (pyarrow and fastparquet) and neither engine supports appending to Parquet files natively due to the structure of Parquet format --- It is columnar and optimized for immutable, efficient batch writes, not row-wise appends
### So is there a workaround to this problem?
- According to what I found:
  - Read the old file.
  - Concatenate the new data
  - Overwirte the file
### But if I have to do this will this be called schema evolution?
- No, manually reading, concatenating, and overwriting Parquet files is not true schema evolution. It’s a workaround, not a native solution.
- Schema evolution means the system automatically handles changes in the data structure over time such as:
  - Adding new columns
  - Removing or renaming columns
  - Changing data types
- And still allows:
  - Reading old and new data togeather
  - Appending new schema-compaint records
  - Querying without breaking due to version mismatches
- This is supported in systems like:
  - Apache Hive
  - Apache spark
  - Delta Lake
  - Apache Iceberge
  - BigQuery
  - SnowFlake
### Colcusion I have to handle schema evolution manually in pandas.
- Schema evolution: If I add a new column age and aligned old data with it
- Old rows have Nan or pd.Na for missing values.
- This way I will have to avoid errors and preserve backward compatibility.
### Best Practices:
- Always read + align schema before appening.
- Use a schema versioning strategy if your dataset evolve frequently
- Store metadata alongside your Parquet files it needed.
- I have to consider Delat Lake, Hudi or Iceberge if i want an automatic schema evolution support.

### Re-trying schema evolution manually in pandas 

````python
df_day1 = pd.DataFrame({
    "id" : [1,2],
    "name" : ["Rollex","Ballistic"]
})
df_day1
````

#### Saving the dataFrame in Parquet file format

````python
df_day1.to_parquet("day1_data.parquet")
````

#### Read the data from parquet and convert it to the dataframe

````python
parquet_dataFrame = pd.read_parquet("day1_data.parquet")
parquet_dataFrame
````

#### Add a new column in the dataFrame generated from the stored paraquet file 

````python
df_day2 = pd.DataFrame({
    "id":[3],
    "name":["Ignition"],
    "roll_no":6000
})
df_day2
````

````python
df_old = parquet_dataFrame

# Align schemas (schema evolution logic)
# Already has it 
df_day2["age"] = df_day2.get("age",pd.NA)
# Add missing columns to the old data_frame
df_old["age"] = pd.NA

# combine and overwrite
df_combine = pd.concat([df_old,df_day2],ignore_index = True)
df_combine.to_parquet("day2_data.parquet",index=False)
````

#### df_day2["age"] = df_day2.get("age",pd.NA) Explaination:
- Purpose: Ensure that the "age" column exists in "df_day2".
- How it works:
  - df_day2.get("age",pd.NA) looks for the "age" column in "df_days2".
  - If it exists, it keeps its original values.
  - If it doesn't exist, It fills the column with pd.NA (Pandas' scalar missing value).
- This is the safe way to ensure the column exists without raising an error.
#### df_old["age"] = pd.NA Explaination:
- Purpose: Add the "age" column to the old DataFrame (df_old) and fill it with missing values
- Since df_old doesn't originally have "age", this line adds the column explicitly with all values as missing (pd.NA).

#### Now finally reading the latest dataFrame stored in parquet file. 

````python
df_latest_data = pd.read_parquet("day2_data.parquet")
df_latest_data
````
