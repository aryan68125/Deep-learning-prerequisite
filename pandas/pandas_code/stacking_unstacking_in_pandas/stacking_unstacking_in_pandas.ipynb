## Stacking

Let's say you want to analyze a stock market data, where you have two level of column headers.
- The first level is the metric that you are analyzing ```[price and (P/E)]```
- The second level is the company itself.

<br>

Now there is a way to make the second level header of the dataFrame in this case ```company``` can be transposed and make it a row using ```.stack()``` function. <br>
Similarly you can also make the first level header of the dataFrame in this case ```metrics``` can be transposed and make it a row using ```.stack()``` function. <br>
To stack the innermost level of column headers use ```.stack()``` method on the dataFrame. 

````python
df.stack()
````

### Analysis: 
```
/tmp/ipykernel_19420/3977694045.py:1: FutureWarning: The previous implementation of stack is deprecated and will be removed in a future version of pandas. See the What's New notes for pandas 2.1.0 for details. Specify future_stack=True to adopt the new implementation and silence this warning.
  df.stack()
```
### What does this warning mean? <br>
- Pandas is warning you that the current implementation of .stack() will change (or be removed) in a future version.
- They’ve introduced a new implementation of .stack(), and you can start using it today by adding future_stack=True.

````python
df_stacked_level_1 = df.stack(future_stack=True)
df_stacked_level_1
````

If you want to stack the first level of column headers instead then all you need to do is this ```.stack(level=0)```<br>
There is another argument in this function ```.stack()``` and that is ```dropna``` where it can be used to drop all the NaN values from the dataFrame if required when transposing the header in the dataFrame.

````python
df_stacked_level_zero = df.stack(level=0,future_stack=True)
df_stacked_level_zero
````

## Unstacking

We can use unstack to perform reverse transformation and restore the dataFrame into it's original form.

### Unstack dataFrame where level=1 header was tranposed i.e the ```company``` header was transposed from columns to rows.
Restore dataFrame to it's original form.

````python
df_stacked_level_1.unstack()
````

#### Original dataFrame looks something like this.

````python
df = pd.read_excel("/home/aditya/github/Deep-learning-prerequisite/pandas/dataset/stocks_to_demo_stacking.xlsx",header=[0,1])
df
````

#### Analysis:
As you can see that the dataFrame that we got is not the same as what the original dataFrame actually is

#### Let's try to drop the multi-level column headers called ```Unnamed: 0_level_0	```

````python
restored = df_stacked_level_1.unstack()
droped_restored_df=restored.drop(columns="Unnamed: 0_level_0",level=0)
droped_restored_df
````

#### Huh? It seems that the Company column that had dates in it was under the ```Unnamed: 0_level_0``` header I need to some how switch places with the company that has dates and the company that has NaN values in them in the dataFrame.

#### Ok so instead of trying to drop ```Unnamed: 0_level_0```, let's try to drop the columns that have NaT and NaN values in it. 

````python
restored = restored.dropna(axis=1, how="all")
restored
````

#### Now this dataFrame looks like how the original dataFrame looks like

### Unstack dataFrame where level=0 header was tranposed i.e the ```metrics (price and P/E)``` header was transposed from columns to rows.
Restore dataFrame to it's original form.

````python
df_restored = df_stacked_level_zero.unstack(level=1)
df_restored
````

#### Drop the columns that have ```NaT``` or ```NaN``` values in them.

````python
dropped_restored_df = df_restored.dropna(axis=1,how="all")
dropped_restored_df
````

#### swap the multi-level header move the ```company``` header to level 1 and ```matrics``` to level 0

````python
dropped_restored_df.columns = dropped_restored_df.columns.swaplevel(0,1)
dropped_restored_df
````
