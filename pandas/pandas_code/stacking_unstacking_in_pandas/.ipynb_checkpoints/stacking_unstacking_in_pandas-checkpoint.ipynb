# Stacking and Unstacking in pandas

````python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import itertools
import seaborn

df = pd.read_excel("/home/aditya/github/Deep-learning-prerequisite/pandas/dataset/stocks_to_demo_stacking.xlsx",header=[0,1])
df
````

Let's say you want to analyze a stock market data, where you have two level of column headers.
- The first level is the metric that you are analyzing ```[price and (P/E)]```
- The second level is the company itself.

<br>

Now there is a way to make the second level header of the dataFrame in this case ```company``` can be transposed and make it a row using ```.stack()``` function. <br>
Similarly you can also make the first level header of the dataFrame in this case ```metrics``` can be transposed and make it a row using ```.stack()``` function.

````python
df.stack()
````
