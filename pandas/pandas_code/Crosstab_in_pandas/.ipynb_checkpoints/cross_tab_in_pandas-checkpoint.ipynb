# Crosstab in pandas
```.crosstab()``` function allows you to compute a frequency table of two or more variables, which summarizes the distribution of values in the data and provides insightes into the relationships between the variables. It is a type of table in a matrix format that displays the multivariate frequency distribution of variables. They are heavily used in survey research, business, intelligence, engineering and scientific research. <br>
A variable having two categories is called binary variable. When both variables are binary, the resulting contingency table is a 2 X 2 table. Also, commonly known as a four-fold table because there are four cells. <br>
**Each cell in the table tells you how many items fall into both categories at the same time.**

<br>

| Alcohol consumption | Smoke: Yes | Smoke: No | Total |
| ------------------- | ---------- | --------- | ----- |
| Low                 | 10         | 80        | 90    |
| High                | 50         | 40        | 90    |
| Total               | 60         | 120       | 180   |


<br>

A contingency table can summarize three probability distributions - Joint, marginal and conditional.
- The joint distribution describes the proportion of the subjects jointly classified by a category of X and a category of Y. The cells of the contingency table divided by the total provides the joint distribution. The sum of the joint distribution is 1.
- The marginal distributions describes the distribution of the X (row) or Y (column) variable alone. The row column totals of the contingency table divided by the total provides the joint distribution. The sum of the joint distribution is 1.
- The conditional distributions describe the distribution of one variable given the levels of the other variable. The cells of the contingency table divided by the row or column totals provide the conditional distributions. The sum of a conditional distribution is 1.

<br>

**Interpretation:**
- 10 people drink little alcohol and smoke.
- 80 people drink little alcohol and don’t smoke.
- 50 people drink a lot and smoke.
- 40 people drink a lot and don’t smoke.

<br>

So the table summarizes the joint distribution of the two variables. <br>

If you want to quickly see the pattern (e.g., “Do heavy drinkers smoke more often?”), scanning row by row is painful.
A contingency table condenses raw data into a simple grid so patterns & relationships are visible. <br>

**When is it used?**
- Exploratory Data Analysis (EDA): When you’re trying to understand relationships between two categorical variables.
  e.g. Do smokers tend to drink more? Do males/females prefer different products?
- Feature Engineering in ML:
  - Sometimes categorical relationships matter when building models (like Naive Bayes, Decision Trees).
  - ontingency tables help you understand which categories are related.
- Hypothesis Testing (Chi-square test):
  - Used to check if two variables are independent or related.
  - Example: Is smoking related to alcohol consumption?
- Business/Health decisions:
  - Hospitals: Does a treatment’s success depend on gender?
  - E-commerce: Are product preferences related to age group?

### Use ```margins``` and set it to ```True``` if you want to show the total in the contingency table 
When you set ```margins=True``` then adding totals (sum rows and columns) are also shown in the contigency table.
#### Why use margins?
- To quicky see totals across rows/columns.
- Good for summary reporting (like in Excel pivot tables).
- Useful in EDA when you want both detailed counts + overall totals.
#### When not to use it?
- If you only care about relationships and not totals.
- If you are normalizing percentages (```normalize```) -> margins can be confusing because totals also get normalized.
#### Options for ```margins```
- ```margins=False``` (default) -> no totals.
- ```margins=True``` -> adds an "All" row and columns. where total is shown row and columnwise
- ```margins_name=CustomName``` -> change the label of the total row/col.

#### Create a contigency table between Nationality vs Handedness also show total 

````python
pd.crosstab(df["Nationality"],df["Handedness"],margins=True)
````

#### Create a contingency table between Nationality vs Sex also show total

````python
pd.crosstab(df["Nationality"],df["Sex"],margins=True)
````

#### Create a contigency table between Nationality vs Age also show total

````python
pd.crosstab(df["Nationality"],df["Age"],margins=True)
````

#### Create a contigency table between Sex vs Handedness also show total

````python
pd.crosstab(df["Sex"],df["Handedness"],margins=True,margins_name="Total")
````

### Passing multiple variables at the column level when generating a contigency table using the ```.crosstab()``` function.

#### Nationality vs [Sex,Handedness] contigency table

````python
pd.crosstab(df["Nationality"],[df["Sex"],df["Handedness"]],margins=True,margins_name="Total")
````

#### Nationality vs [Age,Handedness] contigency table

````python
pd.crosstab(df["Nationality"],[df["Age"],df["Handedness"]],margins=True)
````

### Use ```normalize="index"``` to calculate and show percentage in the generated contigency table in Pandas
By default ```pd.crosstab``` function shows the counts. ```normalize``` lets you see proportions (fractions/percentages) instead.
#### Why use ```normalize```?
- To compare **proportions** instead of raw counts.
- Useful when groups are uneven in size (example, if you have 10000 males but only 50 females, raw counts mislead, but percentages make sense)
- Crucial in ML feature exploration -> helps check if categories are biased toward certain outcomes.
#### When not to use it?
- When absolute counts are important (example: "How many customers bought product X?")
- In very small datasets (percentages can exaggerate differences).
#### Options for ```normalize```
- ```normalize=False``` (default) raw counts
- ```normalize=True``` normalize over the whole table (fractions of grand total)
- ```normalize="all"``` same as when ```normalize``` was set to ```True```
- ```normalize="index"``` normalize row-wise (each row sums to 1)
- ```normalize="columns"``` normalize column-wise (each column sums to 1)

#### Generate a contingency table between Nationality vs Handedness, calculate the percentage and show it in the generate dataFrame.

````python
pd.crosstab(df["Nationality"],df["Handedness"],margins=False,normalize="index")
````

#### Generate a contingency table between Nationality vs Sex, calculate the percentage and show it in the generate dataFrame.

````python
pd.crosstab(df["Nationality"],df["Sex"],margins=False,normalize="index")
````

#### Generate a contingency table between Sex vs Handedness, calculate the percentage and show it in the generate dataFrame

````python
pd.crosstab(df["Sex"],df["Handedness"],margins=False,normalize="index")
````

## NOTICE : In AI/ML context
- **Margin:** helps you understand dataset balance (example: total positives vs negatives).
- **Normalize:** helps you see probabilities -> essential in ML when checking for class imbalance for feature distribution.
