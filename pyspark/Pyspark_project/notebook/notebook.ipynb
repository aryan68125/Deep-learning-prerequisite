# Pyspark project
## OverView of this project : 
![diagram](images/diagram.png)
- We have entity data available in the HIVE tables. The data ingestion pipeline is already in place.
- We are supposed to develope a spark application that does the following.
    - Take load date as an input argument.
    - Read entity data from the hive tables for the given load date.
    - Process and prepare data according to the requirements.
    - Sends prepared entity to Kafka topic.
- Follow the best practices, such as modular design, code reuse, unit testing etc...
- Estimate computation requirements, such as the number of executors, CPU cores memory etc.
- Work with the devops team to define CI/CD pipeline.
