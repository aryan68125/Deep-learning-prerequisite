# Pyspark project
## OverView of this project : 
![diagram](images/diagram.png)
- We have entity data available in the HIVE tables. The data ingestion pipeline is already in place.
- We are supposed to develope a spark application that does the following.
    - Take load date as an input argument.
    - Read entity data from the hive tables for the given load date.
    - Process and prepare data according to the requirements.
    - Sends prepared entity to Kafka topic.
- Follow the best practices, such as modular design, code reuse, unit testing etc...
- Estimate computation requirements, such as the number of executors, CPU cores memory etc.
- Work with the devops team to define CI/CD pipeline.
## Requirements : 
- We have entity data in the hive tables.
- We are asked to read data from the HIVE tables
- Transform the data according to the requirements.
- After transformation send the transformed data to Kafka.
- Let's talk about the data and the transformation requirements :
    - You are asked to read the raw data about the entities and prepare it for sending to kafka
    - Let me explain the requirements for one entity.
        - We have an account entity.
        - The raw data for the account entity is exported into three HIVE tables.
        - Here is the list of the tables : 
            - Account
            - Parties
            - Party Address
        - The account entity is made up of combining data from all three tables.
        - These three tables are for one entity. Here we have a few 100 entities so the complete project for moving all entities is an enormous requirement.
        - But here I am only going to work with one entity and the same approach is applied on the other entities in the bank.
