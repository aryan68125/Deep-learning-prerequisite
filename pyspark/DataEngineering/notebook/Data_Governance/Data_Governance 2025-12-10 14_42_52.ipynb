# Data Governance Overview
Data governance refers to the process of managing the availability, usability, integrity and security of data.

Data governance has become extremely important in today's data-driven landscape and is needed to establish trustworthy data that is readily accessible to those who need it in accordance with internally defined policies or external regulatory requirements.

When we talk about data governance, we can generally divide the discussion into four main areas.

## Data Access Control : 
Prganizations need to be able to lock down data and data-engineering artifacts such as files, tables and machine learning models. Permission to access should be granted only to those who need it.

## Data Access Auditing : 
Any data governance program requires knowledge of how data is being used and that doesn't just mean who is accessing the data it also means when and how it's being used.

## Data Lineage : 
The journey your data takes from its origin as it moves through your pipelines is reffered to as data lineage, and being able to trace it is very beneficial in a number of ways
  - First, it's much easier to identify the root causes of processing issues.
  - It enables you to predict the impact of proposed upstream changes on your processing.
  - It helps compliance teams prove that data and reports come from trusted and verifiable sources
  - In short, it fuels a better understanding of the data for data analysts scientists and engineers.

## Data Discovery : 
Locating data is always a challenge, particularly in a data lake but a data governance program requires a readily accesible inventory of data assets that can be easily searched.

## Challenges in the DataLake
Since the emergence of the data lake paradigm, data governance in this context has been complex, perhaps unnecessarily so.

Many organizations employ a data lake alongside a data warehouse, which leads to data duplication and hence, data drift.

Governance happens at two different levels across these platforms, both of which often employ fundamentally different tool stacks.

This makes it hard to collaborate and control things consitently

## Governance for data, analytics and AI is complex
Data and AI governance today is complex. We have all these different personas on the left that we are looking to serve and all these different types of technology that we are trying to use to serve them and this creates a really complex environment for governance.

In a typical enterprise today, you have lot of data shared in data lakes. This simply be for example - AWS S3 bucket
- To control permissions on data in the data lake you set permissions on files and directories. This means you cannot set fine-grained permissions on rows and columns because governance controls are at the file level, data teams must carefully structure their data layout to support the desired policies. For example, a team might partition data into different directories by country and give access to each directory to different groups but what should the team do when governance rules changes? If different regulations exists inside one country, the organization may need to restrucutre all its directories and files it becomes extremely complex to rewrite all the policy changes 
- With most data lakes, you not only have files but you also have metadata. for example, you might have a hive metastore that keeps track of table definitions and views. So you have to give permission on tables and views, in addition to file-level permissions. This can cause sync with underlying data. There is no guarantee that is users have access permissions on files, they will have corresponding permissions on table that are made up of data from those files or vice versa.
- Then you have your data warehouse, where permissions are more fined-grained on tables, columns and views. But, this is yet again, a different governance model, and now you have created data silos with data movement across two systems, each with a different governance model. As a result, your governance permissions, conduct audits or discovered and share data.
- Now, add other types of assets like dash boards, machine learning models, and notebooks. Each have their own permission models and tech stacks. The problem gets bigger when your data assets exist across multiple clouds with different access management solutions.
- What you need is a unified approach to simplify governance for data and AI.

## Metastore
The metastore is the top-level logical container in Unity Catalog. It's a construct that represents the metadata, that is, information about the objects being managed by the metastore as well as the access control lists that givern access to those objects.

Don't confuse the Unity Catalog metastore with the Hive metastore. The HIVE metastore which will be familiar to existing databricks users is the traditional default metastore linked to each Databricks workspace and while it may seem functionally similar to a Unity Catalog metastore, Unity Catalog metastore offers improved security and auditing capabilities as well as an connect to data to data in a hive metastore a little later on.

It's best to think of a metastore as a logical construct for organizing your data and its associated metadata, rather than a physical container itself. This theme is reflected in the physical makeup of a metastore and accepting this idea will also make some of our best practices more sensible, which we will cover later in this session.

The metastore essentially functions as a reference for a collection of metadata and a link to a cloud storage container.

The metadata that is information about the data objects (like a table's columns and data types for example), as well as access control lists for those objects is stored in the control plane. This ties a metastore to a cloud region. This is why account adminitrators are promted for a region when creating a metastore.

Data related to objects managed by the metastore is stored in a cloud storage container in the same region, which is also configured as part of the metastore setup.

We don't dive much further into these details at the present time , through we will revisit these concept throughout this session where they apply.
