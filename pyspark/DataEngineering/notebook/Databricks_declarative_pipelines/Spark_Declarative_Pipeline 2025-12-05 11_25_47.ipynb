**NOTE :** Later in this course, we will dive eeper into fundamentals of streaming tables and materialized views. For now, focus on understanding the overall architecture and how data incrementally flows through these data types in the pipeline.

### OverView of incremental Processing in Declarative Pipelines (Run 1):
In the first run of this pipeline, raw data will be ingested from the data source and appended to the bronze streaming table.

Next, the pipeline transforms the new data from the bronze layer according to your transformation logic and appends the results to the silver streaming table.

Finally, the materialized view processes the data at the gold level. Depending on various factors, this processing can happen incrementally or as full refresh, ensuring your gold-level data is always up to date and ready for downstream use. Materialized views contain a variety of built in optimizations then run on serverless.

### OverView of incremental processing in declarative pipelines (Run 2):
When the pipeline runs a second time, it uses checkpoints to track what data has already been ingested and processed (auto-loader). In this senario, let's imagine new data has arrived in the raw data source.

In this case, only the new data is incrementally ingested from the data source and appended to the bronze streaming table. The pipeline ignores data it has already processed to avoid processing overhead from the previously ingested data.

Next, only the new data from the bronze table is incrementally transformed and appended to the silver streaming table, ignoring data that has already been transformed in thie first run,  again avoiding the processing overhead from the previous transformation.

Finally, the materialized view processed the full silver streaming table at the gold level, Depending on various factors, this processing can be incremental or a full refresh, ensuring your gold-level data stays accurate and up to date for your downstream consumers.

This incremental processing approach improves efficiency by only handling new data that hasn't been processed before. It avoids unecessary reprocessing of existing data. Materialized views add even more value by providing built-in optimization, making it faster and more efficient to aggregate or query the streaming tables.

## Dataset Types Overview
### Materialized View
- Each time a materialized view is updated, **query results are re-calculated to reflect changes in upstream datasets**
- These views are created and maintained automatically by the pipeline, so you don't have to manage refresh logic.
- Created and kept up-to-date by pipeline.
- Use the ```CREATE OR REFRESH MATERIALIZED VIEW``` syntax.
- While materialized views are commonly used in the gold layer, they can be used anywhere in your pipeline when performance and reusability matter.
- **Where applicable**, results are **incrementally refreshed for materialized views,** avoiding the need to completely rebuild the materialized view when new data arrives (Serverless ONLY)
- **Incrementally refresh for materialized views** is a cost-based optimizer o power fast and efficient transformations for materialized views on Serverless compute.
- Behind the scene a const-based optimizer determines wheather to incrementally refresh or fully re-compute the view. Ensuring fasrt and efficient performance with minimal resources usage.
- **Create materialized view with sql:**
  ```sql
  CREATE OR REFRESH MATERIALIZED VIEW 3_gold_db.gold_orders_by_date AS 
  SELECT date(order_timestamp) AS order_date, count(*) AS total_daily_orders FROM 2_silver_db.order_silver GROUP BY (order_timestamp);
  ```
  **Important NOTE:** Where ever possible,  the system will use incremental refreshes to update the view efficiently, rather than re-building it from the scratch. This is supported in serverless compite and driven by a cost-based optimizer for performance.
  This approach is ideal for creating gold-layer aggregations or business-ready outputs with minimal overhead and high performance.
### Views
Views create virtual tables that do not store any physical data, instead, they are simply logical representations based on the sql query defined in your pipeline.

There are two types of views you will commonly work with 
- TEMPORARY VIEW : 
  - Temporary views are short-lived that exist only for the duration of the pipeline run and are private to the pipeline that defines them.
  - These views are not registered as objects in Unity Catalog, so they remain internal and isolated. 
  - You create them using the SQL statement: CREATE TEMPORARY VIEW
  - Temporary views are especially useful for intermediate queries or calculations that support your pipeline logic but don't need to be exposed to end users or other pipelines.
- VIEW
  - Views construct a virtual table based on the result set of a SQL query, but they do not store any physical data.
  - Unlike temporary views, these views are registered as objects in Unity Catalog, making them available and managable across your environment.
  - To create them, use the SQL statement ```CREATE VIEW```
  - Important Limitations:
    - The pipeline must be a Unity Catalog pipeline to create these views.
    - Views cannot contain streaming queries and cannot be used as streaming sources within a pipeline.

## Ensure Data Quality with Expectations
### Creating Expectations
LakeFlow Spark Declarative Pipelines lets you apply data quality rules called expectations during your ETL processes. These rules validate your data row-by-row to ensure integrity.

The general syntax is: ```CONSTRAINTS <constraint-name> EXPECT (<column condition>) [ON VIOLATION <action>]```

You can specify three types of actions when a violation occurs:
**1. WARN(default):**
  - Logs a warning but still writes invalid rows to the target.
  - Example: to warn if the notifications column contains values other than 'Y'
  or 'N': CONSTRAINT valid_notification EXPECT (notification IN ('Y','N'))
  - Logs will include counts of valid vs. invalid records and other metrics, helping you monitor data quality without stopping the pipeline.

**2. DROP :**
  - Drops invalid rows from the output.
  - Example: To drop rows where order_timestamp is before "2021-01-01":
    ```CONSTRAINT valid_date EXPECT (order_timestamp > "2021-01-01") ON VIOLATION DROP ROW```
  - Invalid rows are excluded, and counts of dropped rows are logged for auditing.

**3. FAIL :**
  - Fails the specific flow if the rule is violated, requiring manual intervention. Other flows in the pipeline continue running.
  - Example: To fail if customer_id is NULL: ```CONSTRIANT valid_id EXPECT (customer_id IS NOT NULL) ON VIOLATION FAIL UPDATE```
  - This helps catch critical issues that must be resolved before proceeding.

**Adding constraints in declarative pipelines:**

Here we will create a streaming table named orders_silver. This streaming table will have the following three constraints : 
  - WARN - ```CONSTRAINT valid_notifications EXPECT (notifications IN ('Y','N'))```
  - DROP - ```CONSTRAINT valid_id EXPECT (customer_id IS NOT NULL) ON VIOLATION DROP ROW```
  - FAIL - ```CONSTRAINT valid_date EXPECT (order_timestamp > "2021-01-01") ON VIOLATION FAIL UPDATE```

Let's see how the Expectations are applied within the pipeline architecture <br>
![expectation_in_ETL_pipeline](images/expectation_in_ETL_pipeline.png) <br>
Here's what happens step by step :
- A row of data enters the pipeline 
- That row is evaluated against any defined expectations (constraints).
- If the row passes all expectations, it's kept, and the pipeline continues processing as normal.
- If the row fails an expectation, the action defined in the constraint determines what happens next:
  - WARN (default): the failure is logged is still included, and the pipeline continues.
  - DROP : The row is discarded, but the pipeline continues processing remaining data
  - FAIL : The pipeline fails immediately for that specific flow, and manual intervention is required. Other flows are unaffected.

## Streaming joins overView : 
### Stream-Snapshot Join OverView
When performing joins with streaming tables in declarative pipelines, it's important to understand the different join types and how they behave.

Let's start with the most straightforward: Joining a Streaming Table to a Static Table (sometimes called a Stream-Snapshot Join or Stream-static join depending on the senario)

In this pattern, we incrementally join new data from a streaming table to a static lookup table to create another streaming table.

For example: imagine you are ingesting a stream of transactions that includes a country_code column. You can join that with a static reference table that maps each code to a full country name.

This is useful when you want to enrich your streaming data with reference information that doesn't change often.

As the new data arrives in the streaming table, It is joined in real time with the static reference table.

That means every new record in the streaming table is matched against all available data in the static table, ensuring a complete and accurate join each time.

The result of that join is then appended to your final streaming output table, continuing the pipeline flow.

This is a great pattern for enritching streaming data with context, like converting country codes, product IDs, or user roles into readable formats using static reference data.

It's efficient, reliable and works well since only one side of the join, the streaming table is changing over time.

A key detail to understand here is that as new data is appended to your source streaming table, only the new rows are joined with the entire static lookup table.

This is an incremental join, meaning the static table doesn't need to be reprocessed, only the incoming streaming data is evaluated.

This keeps the process highly efficient and ensures that each new record is immediately enriched with the latest reference data from the static table.

### Joining Streaming Table with a Materialized View
Another supported pattern is joining two streaming tables using a materialized view.

The goal here is to join all rows from both streaming tables each time the pipeline is run.

This type of join is typically used when both input datasets are changing continuously and you need to combine them regularly to produce a unified up-to-date result.

For example : imagine one stream contains customer activity and another contains product catalog updates. You could join them using a materialized view to enrich activity data with the most recent product infromation.

Because both sides are streaming, a materialized view is required to handle this join efficiency and keep the results current.

The materialized view will process all new rows from both tables and incrementally refresh depending on pipeline configuration and compute mode.

As the new data is appended to both streaming tables, the materialized view detects the changes and responds accordingly.

Each time the pipeline runs, the materialized view will efficiently compute the join by processing all the current data from both streaming tables.

This ensures that every new record from either side is matched correctly, producing an up-to-date and complete output.

The use of a materialized view allows the join to scale efficiently, leveraging incremental refresh where possible to avoid unnecessary recomputation.

This join pattern is ideal when both data seources are live and frequently updated and you want to keep your results synchronized across both.

As new data is added to the streaming tables , the materialized view will again efficiently compute all the data in the streaming tables and join the rows.

### The Stream-Stream joins
The final type of streaming join is a stream-stream join, which is designed to incrementally join new data from two streaming tables as it arrives.

In this pattern, only the new incoming data from each stream is joined, past data is not considered during each run.

## Change Data Capture OverView
### What is change data capture?
It is a foundational concept for keeping data in sync across systems.

Change Data Capture (CDC) is a technique used to track and capture changes in a data source (such as a database, lakehouse or data warehouse). Those changes are then applied to a target table, for example : your Lakehouse , to keep it up to date with the latest state from the source.

CDC is also closely related to how we handle slowly changing dimensions or SCDs which define how historical changes are tracked and stored in your target. There are two main types of SCD we will focus on:
- **SCD type 1:** Overwrites existing data (not history tracking)
- **SCD Type 2:** Tracks historical changes storing previous versions of records

Imagine we are working with a customer table.
- Our source data contains new customers records as well as updates and deletes to existing customers.
- We want to apply those changes to our target table using either SCD type 1 or SCD type 2 logic (We will dive deeper into what those types mean and how they are implemented shortly) to keep our target customers table up to date with the latest information.
### Showing changing dimensions SCD Type 1:
**SCD Type1 Overview (Target table is overwritten with the latest values)**
- When a record is updated by it's key(s), the existing row is overwritten in the target table and no version history kept.
- When a record is delete by it's key(s), it is removed from the target table and no version history is kept.
- Only the current state of each record is retained in the target table, no tracking of previous values or changes.

When we apply SCD Type 1, our target table is updated with the latest customer information, without keeping any historical versions of the data based on the **CustomerID** (ID column) and **ProcessDate** (sequence column)

Let's walk through what happens in this example: <br>
![SCD_type_1](images/SCD_type_1.png) <br>
- Peter (customer_id 1): His address is updated to the latest address based on the ProcessDate to 123 Main St. from the updates table.
- Samarth (customer_id 2): He's been deleted, so his row is removed from the target table.
- Kostas (customer_id 3): He's a new customer, so his record is inserted into the table.

The end result is that the customers table contains a current snapshot of all active customers, with no history, just the most recent customers.

This is the simplest CDC strategy, and it's ideal when maintaining historical changes isn't necessary. You just need the latest, most accurate data.

### Showing changing dimensions SCD Type 2:
![SCT_type_2](images/SCT_type_2.png)
- When a record changes 
  - The old record is preserved with an additional column indicating its validity period (start date, end date or current flag) and a snew record is inserted with the updated information.
  - When a record is marked as deleted, the record is kept and a column indicates the record is inactive.
- SCD Type 2 is used when historical data is important and the system needs to track how attribute changes over time (like tracking changes in a customer's address or status)
  - **_START_AT :** shows when the row became active
  - **_END_AT :** shows when the row became inactive (if it has)
    - A null **_END_AT** means the row is currently active.
 
Let's breakdown the above example as shown in the picture
- **Customer_id 1 :** Peter
  - Two records exists for Peter:
    - The active record shows Peter's updated address.
      - **_END_AT** is null -> still active
      - **_START_AT** marks when the update occured
    - The inactive record holds Peter's old address. We can tell it inactive because **_END_AT** is populated --> no longer current.
- **Customer_id 2:** Sarmath
  - Since Samarth deleted his account hist existing record is now inactive.
  - A date is added to **_END_AT** to indicate when he was removed.
- **Customer_id 3:** Kostas
  - A new customer, so his record was inserted into the table.
  - **_START_AT** shows when he joined
  - **_END_AT** is null --> the record is currently active

LakeFlow provides a built-in declarative CDC mechanism using AUTO CDC INTO statement.

You use AUTO CDC INTO to handle updates, inserts and deletes in your pipeline.
Here's what each part of the statement does: <br>
![using_auto_cdc](images/using_auto_cdc.png) <br>
- AUTO CDC INTO customers - Applies all CDC events to the target table named customers.
- FROM STREAM updates specifies the source stream that contains CDC events (inserts, updates, deletes).
- KEYS (Customer_id) Defines the primary key used to match rows between source and target.
- AUTO CDC AS DELETE WHEN operation = "DELETE" defines when a row should be treated as a delete, based on the operation column.
- SEQUENCE BY PROCESSDATE : Establises the event order , critical for apllying changes in the correct sequence
- COLUMN * EXCEPT (operation) Includes all the columns except the CDC operation indicator in the target.
- STORED AS SDC TYPE 2
  - Stores the history of changes using the SDC Type 2 pattern 
  - If ommited, the default is SDC TYPE 1 (Overwrite only).

```python
# Delta Live Tables (DLT) incremental ETL pipeline
# File: dlt_incremental_pipeline.py
# Purpose: Ingest customers.csv, orders.csv, sales.csv incrementally using Auto Loader (cloud_files),
#         clean & enforce schema, and produce joined gold tables (customer_orders, customer_sales).
# Instructions:
# 1. Create a new Delta Live Tables pipeline in Databricks UI (New -> Delta Live Tables).
# 2. Set "Pipeline mode" to "Continuous" (or "Triggered" if you prefer runs triggered on demand).
# 3. Point the "Notebook path" or upload this Python file as the pipeline source.
# 4. Set the storage root for pipeline checkpoints and schema inference (e.g. dbfs:/mnt/dlt/).
# 5. Add any needed pipeline libraries and cluster configuration.

import dlt
from pyspark.sql.functions import col, to_date, trim
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, FloatType, DateType

# -----------------------------------------------------------------------------
# Configurable paths - update these to match your environment
# -----------------------------------------------------------------------------
RAW_BASE = "/mnt/raw"                # where new csv files arrive (directories customers/, orders/, sales/)
SCHEMA_LOCATION = "/mnt/dlt/schema"  # cloud_files schema location

# -----------------------------------------------------------------------------
# Bronze tables: incremental ingestion using Auto Loader (cloud_files)
# -----------------------------------------------------------------------------

@dlt.table(name="bronze_customers", comment="Raw incremental customers from CSV (Auto Loader)")
def bronze_customers():
    return (
        spark.readStream
            .format("cloud_files")
            .option("cloudFiles.format", "csv")
            .option("header", "true")
            .option("inferSchema", "false")
            .option("cloudFiles.schemaLocation", f"{SCHEMA_LOCATION}/customers")
            .option("cloudFiles.maxFilesPerTrigger", 100)
            .load(f"{RAW_BASE}/customers/")
    )

@dlt.table(name="bronze_orders", comment="Raw incremental orders from CSV (Auto Loader)")
def bronze_orders():
    return (
        spark.readStream
            .format("cloud_files")
            .option("cloudFiles.format", "csv")
            .option("header", "true")
            .option("inferSchema", "false")
            .option("cloudFiles.schemaLocation", f"{SCHEMA_LOCATION}/orders")
            .option("cloudFiles.maxFilesPerTrigger", 200)
            .load(f"{RAW_BASE}/orders/")
    )

@dlt.table(name="bronze_sales", comment="Raw incremental sales from CSV (Auto Loader)")
def bronze_sales():
    return (
        spark.readStream
            .format("cloud_files")
            .option("cloudFiles.format", "csv")
            .option("header", "true")
            .option("inferSchema", "false")
            .option("cloudFiles.schemaLocation", f"{SCHEMA_LOCATION}/sales")
            .option("cloudFiles.maxFilesPerTrigger", 200)
            .load(f"{RAW_BASE}/sales/")
    )

# -----------------------------------------------------------------------------
# Silver tables: clean, cast types, basic validation
# -----------------------------------------------------------------------------

@dlt.table(name="silver_customers", comment="Cleaned customers with enforced schema", table_properties={"pipelines.autoOptimize.managed": "true"})
@dlt.expect_or_drop("valid_customer_id", "customer_id IS NOT NULL")
@dlt.expect_or_drop("valid_signup_date", "signup_date IS NULL OR signup_date RLIKE '\\d{4}-\\d{2}-\\d{2}'")
def silver_customers():
    raw = dlt.read("bronze_customers")

    # Normalize column names (trim) and cast types
    c = (
        raw.select([trim(col(c)).alias(c) if isinstance(c, str) else col(c) for c in raw.columns])
    )

    # Best-effort casts (adjust column names/formats to your CSV)
    out = (
        c
        .withColumn("customer_id", col("customer_id").cast(IntegerType()))
        .withColumn("customer_name", col("customer_name"))
        .withColumn("email", col("email"))
        .withColumn("signup_date", to_date(col("signup_date"), "yyyy-MM-dd"))
    )

    return out

@dlt.table(name="silver_orders", comment="Cleaned orders with enforced schema")
@dlt.expect_or_drop("valid_order_id", "order_id IS NOT NULL")
def silver_orders():
    raw = dlt.read("bronze_orders")

    c = raw

    out = (
        c
        .withColumn("order_id", col("order_id").cast(IntegerType()))
        .withColumn("customer_id", col("customer_id").cast(IntegerType()))
        .withColumn("order_date", to_date(col("order_date"), "yyyy-MM-dd"))
        .withColumn("order_amount", col("order_amount").cast(FloatType()))
        .withColumn("order_status", col("order_status"))
    )

    return out

@dlt.table(name="silver_sales", comment="Cleaned sales with enforced schema")
@dlt.expect_or_drop("valid_sale_id", "sale_id IS NOT NULL")
def silver_sales():
    raw = dlt.read("bronze_sales")

    c = raw

    out = (
        c
        .withColumn("sale_id", col("sale_id").cast(IntegerType()))
        .withColumn("customer_id", col("customer_id").cast(IntegerType()))
        .withColumn("sales_date", to_date(col("sales_date"), "yyyy-MM-dd"))
        .withColumn("sales_amount", col("sales_amount").cast(FloatType()))
    )

    return out

# -----------------------------------------------------------------------------
# Gold tables: incremental joins and aggregations
# Note: DLT supports streaming joins between DLT tables; joins will be incremental
# -----------------------------------------------------------------------------

@dlt.table(name="gold_customer_orders", comment="Joined customers + orders (incremental)", table_properties={"delta.enableChangeDataFeed": "true"})
@dlt.expect("non_null_customer", "customer_id IS NOT NULL")
def gold_customer_orders():
    customers = dlt.read("silver_customers")
    orders = dlt.read("silver_orders")

    joined = (
        customers.join(orders, on=["customer_id"], how="inner")
        .select(
            customers.customer_id,
            customers.customer_name,
            customers.email,
            orders.order_id,
            orders.order_date,
            orders.order_amount,
            orders.order_status
        )
    )

    return joined

@dlt.table(name="gold_customer_sales", comment="Joined customers + sales (incremental)")
def gold_customer_sales():
    customers = dlt.read("silver_customers")
    sales = dlt.read("silver_sales")

    joined = (
        customers.join(sales, on=["customer_id"], how="inner")
        .select(
            customers.customer_id,
            customers.customer_name,
            customers.email,
            sales.sale_id,
            sales.sales_date,
            sales.sales_amount
        )
    )

    return joined

# -----------------------------------------------------------------------------
# Expectations, monitoring & notes
# -----------------------------------------------------------------------------
# - dlt.expect_or_drop: will drop rows that fail validation (you can use dlt.expect to flag instead)
# - Adjust date formats if your CSV uses different date/time formats (e.g. dd/MM/yyyy)
# - For better performance, tune "cloudFiles.maxFilesPerTrigger" and pipeline cluster size
# - Configure pipeline libraries/credentials to access the RAW_BASE path (e.g. mount storage or use S3 credentials)
# - Schema drift: DLT + cloud_files will place evolving schemas under SCHEMA_LOCATION; review regularly
# - To enable updates/upserts in gold tables, consider using dlt.apply_changes with a unique key

# Example of a change table using dlt.apply_changes (uncomment and adapt if you want upserts):
# @dlt.table(name="gold_customer_orders_upsert")
# def gold_customer_orders_upsert():
#     changes = dlt.read("silver_orders")
#     target = dlt.read("silver_customers")
#     merged = dlt.apply_changes(target = "gold_customer_orders_upsert",
#                                source = changes,
#                                keys = ["order_id"],
#                                column_mapping = {"customer_id":"customer_id", "order_amount":"order_amount"})
#     return merged
```
