```python
import os
import json
is_job = False
try:
    raw = dbutils.jobs.taskValues.get("join_customer_and_sales_data", "metadata", None)
    print("Running in Job?", True)
    print(f"raw ---> {raw}")

    # get data from the json output of the job ingest_customers_data
    meta = json.loads(raw)
    catalog          = meta.get("catalog")
    schema           = meta.get("schema")
    table_name       = meta.get("table_name")
    is_job = True
except:
    is_job = False
```

```python
if is_job:
    customers_sales_df = spark.sql(f"""
            SELECT DISTINCT * FROM {catalog}.{schema}.{table_name}
            """)
    customers_sales_df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.{table_name}")
    customers_sales_df = spark.sql(f"""
            SELECT * FROM {catalog}.{schema}.{table_name}
            """)
    display(customers_sales_df)
else:
    print(f"Notebook is running in interactive mode skipping processing duplicates")
```
