# Introduction to Big data and Hadoop
## Business Data Processing Beginning and Evolution
- **COBOL** was the first serious attempt towards data processing and It was designed in 1959.
- The **ORACLE database** achieved the subsequent major success enabling data processing and oracle was founded in 1977.
- Data processing has always been at the center of the software industry. Data processing is an evergreen field in software engineers. Software may evolve but the requirement of processing larger and larger data at faster speed will become more and more critical.
## How do we create data processing applications?
- We have used **RDBMS** technology for many decades.
- Some popular RDBMS systems are ORACLE, SqlServer, PostgreSql, MySql, Teradata and Exadata. They all come with different capabilities but the core technology is the same.
- These RDBMS systems offered us three main features to help us develope data processing applications:
  - SQL : An easy data query language.
    - We use sql for querying data.
  - Scripting languages such as PL/SQL and transact SQL.
    - We use scripting languages to do things that we couldn't do using sql.
  - Interface for other programming languages like JDBC and ODBC
    - They provided JDBC and ODBC so that we could interact with data using the programing languages.
- Things were perfect we could create data processing applications using the above technologies but then the data itself started to evolve.
- In the begining we saw data in rows and columns.
  - **COBOL** stored and accessed this data in plain text files example : CSV files.
  - RDBMS like ORACLE stored this data in a more advanced file format like dbf files. We do not care about the file format used by RDBMS but we see data in the format of rows and columns.
  - With time we developed some new data formats such as JSON and XML. RDBMS was not ready to store and process these JSON and XML files. They needed us to define a row column structure and then store it in the database.
  - However many RDBMS added this capability over some time and started to allow storing and processing JSON and XML files but by that time we developed some new formats example : TEXT, DOC, IMAGE, VIDEO (unstructured data)
  - RDMS was designed to handle such exansion of file formats.
  - So basically we developed three categories of data.
    - Structured
      - The structured data comes in a row column table format.
    - Semi-Structured
      - The semi-structured data does not follow the tabular structure. However it has another well defined key value structure such as JSON or XML. These types of data are called semi-structured data because they don't have a tabular structure but they do have some structure as a key value pair.
    - Un-Structured
      - These types of data don't have any definite pattern or structure.
      - The un-structured data comes in a text files, pdfs and other documents, images and video files.
- With the growth of the internet, social media and mobile apps businesses are collecting more semi-structure or un-structure data.
- Structured data is now a small portion of the overall available data within an organization.
- However RDBMs failed to bring the support for the un-structured data
## Big Data Problems:
Modern data processing applications must also be able to handle these problems : 
- Variety
  - Modern data processing applications must be able to handle all types varieties of data 
  - We need methods to store and process structured , semi-structured and un-structured data.
- Volume
  - The overall data volumen is relatively high these days.
  - Companies collect terabytes and petabytes of data in short periods of time and they don't have weeks and months to process such a large amount of data. They want to process such large volumes of data in a matter of hours and minutes.
  - So we need capabilities to handle large volumes of data.
- Velocity
  - The velocity of collecting data is very high.
  - The requirement of processing velocity is also very high
- These three problems combine are also called **Big Data Problem**.
- We had RDBMS systems developed and metured over many decades. However RDBMS was not designed to handle the big data problem.
- So the industry needed a Big Data platform to do the following:
  - Store high volumes of data arriving at a higher velocity.
  - Accomodate structured, semi-structured and un-structured data variety
  - Process high volumes of a variety of data at a higher velocity.

## Approaches to solve Big Data Solution
There are two approaches to solve the big data problem.
- Monolithic Approach :
  - This approach designes one large and robust system that handles all the requirements. TeraData and ExaData are the examples.
  - These two systems only support structured and semi-structured data. So we cannot call them big data systems but they are designed using a monolithic approach.
- Distributed Approach
  - In a distributed approach we take many smaller systems and bring them togeather to solve a bigger problem.
  - This system uses a cluster of computers to work as a single system. The combined capacity of this system can be equal to or even higher than a single monolithic system.
## Comparisions between Monolithic and Distributed approaches:
These two approaches are compared on the following criteria.
- Scalability :
  - Scalability is the ability of a system to increase or decrease the performance in response to the demand.
  - Lets take a look at both the approaches when it comes to scalability : 
    - Lets say we have a 10 terabytes of data stored in both the systems
    - We have 100 concurrent users reading and writing data to the systems.
    - Assume both the systems are working perfectly fine but over time the number of concurrent users and data volumne increased significantly.
    - Here both the systems have limited capacity i.e the CPU, MEMORY and the DISK_SIZE are fixed.
    - As the number of concurrent users and data volume increases these systems will eventually reach their maximum capacity.
    - If we are able to increase the capacity then the system is deemed to be scalable. If we are not able to increase the capacity then the system is deemed to be not scalable.
    - **Monolithic approach** : 
      - Both the approaches are scalable but scaling a monolithic approach is complex.
      - You will have to call the vendor to increase the capacity. It requires coordination with the system vendors.
      - They will bring in more processors increase the tower size and then install them in it.
      - This approach is known as vertical scalability
      - In a vertical scalability you will have to increase the hight of a single system adding more resources to one single system.
      - Hence a monolithic system is vertically scalable.
    - **Distributed systems** : 
      - On the other side the distributed system is also scalable.
      - Scaling a distributed system is as simple as adding new computers to the network.
      - This approach is called horizontal scaling.
      - In horizontal scaling we add new members to the cluster and increase the length of the network.
      - Horizontal scalability takes less time because you can easily bur some new systems.
      - A distributed system is horizontaly scalable.
- Fault Tolerance and High availability
  - Lets look at both the systems.
  - What happens if a cpu burns out?, a network card fails? or the system's motherboard fails?
  - **Monolithic system:**
    - In an event of hardware failure the monolithic system will stop working.
    - It may not tolorate the hardware failure.
    - The duration of time the system is down due to hardware failure the application will not be available to the users.
  - **Distributed system:**
     - If a computer fails in a cluster the other computers remains working.
     - A distributed system can tolerate many failures.
     - A system failure in a cluster will only reduce the capacity but the overall system remains working.
- Cost-Effectiveness :
  - **Monolithic systems:**
    - Monolithic systems are expensive.
    - Scaling them takes a lot of time. So you may have to start with a large machine even though you may need a smaller one.
  - **Distributed systems:**
    - The initial costing of a distributed system is low because initially you can keep the size of the cluster small and the size can be increased later as needed making it more economical.
    - You can also use an average quality machine that you can get at a reasonable price.
    - You can get cloud environment to get machines at rental prices and create your cluster
    - These options make distributed systems more economical and cost effective.
## Analysis : 
- Engineers evaluated both the **Monolithic** and **Distributed** approaches to design a new data processing system and handle Big Data Requirement.
- The evaluation indicate that the distributed systems could be a better choice.
- This is where a new system called **Hadoop** came into existance.
- **Hadoop** was a revolutionary big data processing platform.
- The **Hadoop** platform was designed and developed in layers.
- The core platform layer offered three capabilities.
  - Distributed cluster formation or Cluster Operation System (**YARN**)
  - Data storage and retrieval on the distributed cluster or Distributed storage (**HDFS**)
  - Distributed data processing using Java programming language. (**Map**/**Reduce**) framework
- (**YARN**) Cluster OS : **Hadoop** was developed as an OS for the distributed cluster.
  - We can create a data processing software, install it on one computer and run it there. The computer comes with an operating system.
  - The OS allows you to use computer resources like CPU, MEMORY and DISKS to run your application.
  - If you don't have an OS you cannot run your application on your system.
  - Similarly when you create a cluster of computers you need a cluster OS.
  - The Cluster OS will allow you to use the cluster resources such as CPU, MEMORY and DISK.
  - A cluster OS has more complex work to do, because a CPU, MEMORY and DISK are spread across the computers that are in the cluster. We are not using a single computer where everything is available on a single machine.
  - A cluster is a group of computers and the cluster OS makes it work like a single large computer.
  - And this was the first capability of the hadoop platform.
- (**HDFS**) Distributed storage : **Hadoop** also allowed us to store data and retrieve it back.
  - The data is internally distributed on the hard disk on the cluster. However **Hadoop** offered us a distributed storage system and allowed us to save and read the data file as we do it on a single machine.
- (**Map**/**Reduce**) Distributed computing : The next feature of **Hadoop** is that it allows us to write a data processing application in Java and run it on the cluster.
  - **Hadoop** offered is a new distributed data processing framework known as map reduced framework for processing large volumes of data.
  - **Hadoop** platform was designed to use a cluster of computers as a single large machine.
  - You can read/write and program data processing applications and run them on a **Hadoop** cluster without realising that you are working on a cluster.
  - **Hadoop** simplified the distributed computing for developers.
  - We develope a data processing application on **Hadoop** the same way we develope the applications for a monolithic systems but the program runs on a distributed computing platform and uses parallel processing.
  - **Hadoop** platform was designed and developed in layers. The core layer offered a cluster OS, Distributed storage and Map/Reduce framework.
  - The **Hadoop** community developed many other tools over and above the **Hadoop** core platform.
    - **HIVE** database : We can run SQL queries just like we do in a relational database. We can configure HIVE on the hadoop platform and run SQL queries on the data just like any other RDBMS system.
    - **HBASE** database
    - **PIG** scripting language : HADOOP also offered some scripting languages like apachae pig scripting
    - **SCOOP** data injection tool
    - **OOZIE** workflow tool.

## What is Hadoop and how it works?
**Hadoop** is a data processing platform that offers the following core capabilities.
- **YARN** stands for (**Y**et **A**nother **R**esource Ma**n**ager) it is the **Hadoop**'s cluster resource manager. We also call it **Hadoop** cluster operating system.
  - An operating systems allows multiple programs to be in memory and run simultaneously and this is acheived by sharing resources like CPU, MEMORY and DISK I/O.
  - The **YARN** does the same it allows multiple applications to run on hadoop cluster and share resources amongs the applications.
  - **Hadoop YARN** has three main components:
    - **RM** Resouce Manager
    - **NM** Node Manager
    - **AM** Application Master
  - Let's assume we have 5 computers and we want to make **Hadoop** cluster using these 5 computers
    - We will have to install **Hadoop** on each computer and configure them to make a **Hadoop** cluster.
    - Installing and configuring a **Hadoop** cluster is as simple as installing software on a computer.
    - Assume we have installed and configured **Hadoop** on these 5 computers and these 5 computers form a **Hadoop** cluster.
    - **Hadoop** uses a master slave architecture, so one of the 5 machines will become a master and the remaining will act as a worker node.
    - Configuring **Hadoop** will also install a **YARN Resource Manager** service on the master node. It also installs the **Node Manager** service on all the worker nodes.
    - ![YARN Config](images/yarn_config_on_a_hadoop_cluster.png)
    - The diagram on the above shows the **YARN** configuration on a **Hadoop** cluster.
    - One node acts as a master and runs a **YARN** **Resource Manager** server. The other 4 nodes acts as a worker and runs a **Node Manager** service.
    - The **Node Manager** will send the node status report regularly to the **Resource Manager**.
    - Now we want to run a data processing application on this cluster.
    - For running an application on **Hadoop** you will have to submit the application to the  **YARN Resource Manager**.
    - We have many ways to submit the application to **YARN**
    - Lets suppose you have submitted a JAVA application to **YARN** using a command line submit tool.
    - Now the **Resource Manager** should run this application on the cluster.
    - The **Resource Manager** will request one of the **Node Managers** to start a resource container and run an **Application Master** container.
    - The application that you submitted runs inside the **Application Master** container.
    - And that's all your application starts running inside the **Application Master** container.
    - ![YARN application startup execution](images/YARN_app_startup_process.png)
    - **What is a container?** : It is a set of resources that includes memory and cpu.
    - For example : A **YARN CONTAINER** may be a set of 2 CPU cores and 4 GB of memory. So whenever we think of a **YARN CONTAINER** we should think of a box of some cpu or memory.
    - Now lets assume we submit another application.
    - The **Resouce Manager** will again ask another **Node Manager** to start another **Application Master** container in one of the worker node and this new application will now start running inside that **Application Master** container.
    - Each application in **YARN** runs on different **Application Master** container.
    - If you have 10 applications running in parallel then you will see 10 **Application Master** container on your **Hadoop** cluster.
- **HDFS** (**H**adoop **D**istributed **F**ile **S**ystem) and it allows you to save and retrieve data files in the hadoop cluster.
  - The **HDFS** has two main components.
    - **NN** - Name Node
    - **DN** - Data Node
  - ![HDFS](images/HDFS.png)
  - Lets assume we have 5 computers shown here (we have already installed hadoop on these computers and created a hadoop cluster)
  - We already know that hadoop works on master slave architecture. Where one node will be designated as a master and the rest of the node will be designated as slaves by hadoop.
  - ![HDFS_NN](images/HDFS_nn.png)
  - **Hadoop** will install a **NN** Name Node service in the Master Node and each worker node runs a **DN** Data Node service. The **NN** Name Node with the **DN** Data Node service forms the **HDFS** (**H**adoop **D**istributed **F**ile **S**ystem).
  - The primary purpose of **HDFS** is to allow us to save our data (Files) in this **Hadoop** cluster and then read it whenever required.
  - **How does it work?**
    - Let's assume we want to copy a large data file onto a **Hadoop** cluster.
    - ![HDFS_save_file](images/HDFS_save_file.png)
    - So you will initiate the file copy command.
    - This file copy request will go to the **NN** Name Node.
    - The **NN** Name Node will re-direct the file copy command to one or more **DN** Data Nodes. Lets assume you get redirection for three **DN** Data Nodes
    - The file copy command will split the file into smaller parts and write those parts in the three **DN** Data Nodes.
    - The file copy command will split the file into smaller parts and each part is known as a block the typical size is 128 MB. 
    - The file copy command will write some files to **DN** Data Node 1 , some will be written in **DN** Data Node 2 and the remaining of them will go to **DN** Data Node 3. The main node facilitates this process and keeps track of all the file's metadata.
    - **What is the File metadata?**
      - File name
      - File size
      - File block, block ID, clodk sequence, block location
    - The file is broken and spread on the **DN** Data Nodes and all the information required to re-assemble the files are kept in the **NN** Name Node.
    - **Now lets talk about the read operation. You will initiate the read operation when you want to read a data file.**
    - The read operation request goes to the **NN** Name Node. The Name Node owns all the information that is required for re-assembling the file from the blocks stored at the data node.
    - So the **NN** Name Node will redirect the read operation to the target **DN** Data Nodes.
    - The read api will recieve data blocks from the **DN** Data Node and re-assemble the file. Using the metadata provided by the **NN** Name Node.

      - Reduce part where you will do the summissions or other types of aggregations and consolidations.

## What was there before Map/Reduce and HDFS:
Before the Map/Reduce and HDFS came into existance datawarehouses existed

    - ![datawarehouse](images/datawarehouse.png)

        - **OLTP** systems (**O**nline **T**ransaction **P**rocessing systems) are the source systems from which data is collected, transformed, and loaded into the data warehouse.
    - Then we processed all that data to get business insights and used it to make the correct business decisions. This is precisely the same as what **Hadoop** promised to offer.
- So the advent of **HDFS** and Map/Reduce started challeging these data warehouses in thre critical respect.
    - Hadoop : 
        - Horizontal scaling
        - Low capital investment
        - Exabytes to Petabytes 
        - Structured , Semi Structured and un Structured data
    - Datawarehouse : 
        - Vertical scaling
        - Large capital 
        - Terabytes to Exabytes of data
        - Only structured Data
- The datawarehouse was challenged and we needed to give a name for this new hadoop approach to solving the problem and eventually its called the DataLake
- **How did dataLake work?**
    - Here is how the dataLakes were initially imagined.
    - ![datalake](images/datalake.png)

    - DataLake also allowed us to collect and process huge volumes of semi-structure and un-structured data. This is what we needed for machine learning and AI

    - However the DataLake technology missed two super-critical technologies that data warehousing offered.

    - So we started adopting different architectures for implementing DataLakes.

            - So we basically collected data in the DataLake storage.

            - Finally we connected the reporting and BI with the datawarehouse.
            - Machine learning and AI still worked on the DataLake but BI and reporting again shifted to datawarehouses.
            - So DataLake married DataWarehouse. This is the current architecture for many organizations.
        - Hoever over some timess a lot of things evolved.

## Spark as a unified data analytics and processing platform.
- Apache spark is a distributed data processing framework.
- ![Spark_ecosystem](images/Spark_ecosystem.png)
- This diagram represents the Spark ecosystem. 
- This figure will help us understand how the whole ecosystem is stiched togeather.
- The spark ecosystem is designed in two layers :
    - The bottom layer is the Spark's core layer (**Spark Engine**):
        - ![Pyspark_architecture](images/Pyspark_architecture.png)
        - The spark's core layer itself has got two parts:
            - A distributed computing engine
            - A set of core apis
            - These two things togeather makes the spark ecosystem.
        - ![spark_architecture2](images/spark_architecture2.png)
        - This whole thing runs on a cluster of computers to offer you distributed data processing
        - However Spark does not manage the cluster. It only gives you a data processing framework.
        - ![cluster_manager_spark](images/cluster_manager_spark.png)
        - So you are going to need a cluster manager. People use different names for this cluster manager , it is also called as resource manager or container orchistrator.
        - However they offer similar services to spark and spark is designed to work with both.
        - Spark was initially based on Hadoop's Map/Reduce and it was adopted in the hadoop platform. So the hadoop's **YARN** resource manager is the most commonly used cluster manager for Spark.
        - However you have other options. 
            - Mesos and Spark's stand alone cluster manager
            - Kubernetes : The newer versions of spark is also compatible with kubernetes as a cluster orchistrator.
        - Spark doesn't comes with an inbuilt storage system. It allows you to process the data which is stored in a variety of storage systems. The most popular and cpmmonly used storage systems are : 
            - **HDFS**
            - Amazon S3 
            - Asure DataLake storage
            - Google Cloud Storage
            - Casandra file system
        - So one thing is clear apache spark does not offer cluster management and storage management. All you can do with apache spark is to run your data processing workloads and that part is managed by the spark's compute engine.
        - The Compute engine is responsible for the bunch of things.
            - Breaking your data processing work into smaller tasks.
            - Scheduling those tasks on the cluster for parallel execution
            - Providing data to these tasks.
            - Managing and Monitoring those tasks.
            - Provide you fault tolerance when a job fails.
            - And to do all these the core engine is also responsible for interacting with the cluster manager and the data storage manager.
            - The spark compute engine is the core that runs and manages your data processing work and provides you with the seemless experience.
            - All you need to do is submit your data processing work to Spark and the Spark core will take care of everything else.
    - Then we have the top layer as a DSLs , Libraries and apis (**Spark Core**):
        - ![spark_core_layer](images/spark_core_layer.png)
        - This layer is the programming interface layer that offers you the apis in 4 major languages.
            - Scala
            - Java
            - Python
            - R
        - These are the apis that we use to write the data processing logic during the initial days of apache spark.
        - However these apis were built on resilient distributed data service **RDD**s and people found them a little tricky to learn and use them for their day to day data processing work. Some people still use them today.
        - However these apis are the most complicated way to work with apache spark. They also lack some performance optimization features. The spark creators are recommending to avoid them as much as possible.
        - However these apis can offer you the highest level of flexibility to solve some complex data processing problems.
    - Top mosy layer (Spark SQL, Streaming, Ml lib [Machine learning], GraphX Graph Computation)
        - The top most layer is the prime area of intrest for most Spark developers and data scientist
        - This layer is again a set of libraries, packages, apis and DSL. These are developed by the Spark community over and above the core api.
        - So you will be using these top level apis and DSLs but internally all of those will be using the Spark core apis and unltimately things will go to the spark compute engine.
        - The top most layer is grouped into 4 categories to support 4 different data processing requirements.
        - However this is just a logical grouping and there is no rigid boundary. Things are going to overlap in most of the real life projects.
        - The first group is a set of two things. 
            - Spark Sql :
                - Spark sql allows you to use sql queries to process your data
            - Spark dataFrames or DataSet apis :
                - Spark DataFrame will allow you to use functional programming techniques to solve your data crunching problems.
            - These apis are available in Java, Scala and Python. These two combined will enable you to solve most of the structured and semi-structured data crunching problems.
        - The second group is Spark Streaming Libraries.
            - These libraries allow you to process a continuous and unbounded stream of data.
        - The thrid group is Ml lib.
            - It is a set of libraries specifically designed to meet your machine learning, deep learning libraries and  AI requirements.
        - The fourth group is the graphX 
            - This is the last group which are the set of graph processing libraries.
            - They allow you write the graph processing algorithems using apache spark. 
        - This top most layer helps you solve your data crunching problems and all these are available in multiple languages such as Java, Python, Scala and R.

### Explaination:
- ```inferSchema=True```
    - **Meaning:** Spark will automatically detect data types (like integer, double, string, etc.) for each column by reading a sample of the data.
    - If you leave it False (default): Spark treats all columns as strings (StringType).
- ```header=True```
    - **Meaning:** The first line of your CSV file contains column names, not data.
    - If you set it to False (default): Spark treats the first line as a normal data row.
    - **Effect:** The DataFrame’s column names come from the header row.
- **pyspark** is the Python API for Apache Spark.
It lets you write Spark programs in Python instead of Scala or Java.
- **pyspark.sql** is a module inside PySpark that provides tools for working with structured data — that is, data organized in tables (rows and columns), similar to SQL tables or pandas DataFrames.
- **SparkSession** is the entry point to Spark SQL functionality.
    - It’s an object that lets you:
        - Create DataFrames (distributed datasets similar to pandas DataFrames).
        - Run SQL queries on data.
        - Read data from various sources (CSV, JSON, Parquet, Hive, etc.).
        - Manage Spark configurations and contexts.
- ```spark``` is the SparkSession object.
