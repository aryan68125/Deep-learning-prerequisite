# Working with dataFrame in pySpark
## DataFrame Transformations and Actions (important for an interview perspective)
DataFrame offers a long list of methods however we can classify them into three main groups

- **Actions:** 
    - Actions trigger a spark job and return to the spark driver.
    - ![spark_execution_architecture](images/spark_execution_architecture.png)
    - Spark is a distributed processing system so it implements a driver and executor architecture in other words a master slave architecture.
    - So every spark program runs as a one driver and one or more executor.
    - Actions are dataFrame operations that **kick off a spark job** execution and return to the **Spark driver**.
- **Transformations:**
    - Spark dataFrame transformation produces a newly transformed dataFrame.
    - Transformations are different than Actions so they do not kick of a spark job and they do not return to the driver.
    - They simply create a new dataFrame and return it.
- **Functions/Methods:**
    - DataFrame methods or functions which are not categorized into Actions or transformations i.e everything which are not either Actions or Transformations are functions/methods

````python
from pyspark.sql import SparkSession

# imports related to spark dataFrame transformation
from pyspark.sql.functions import to_date, to_timestamp, expr

#  Create a spark session
spark = SparkSession.builder.appName("DfApp")\
        .config("spark.sql.ansi.enabled", "false")\
        .getOrCreate()

spark_df = spark.read.format('csv')\
    .option('header','true')\
    .option('inferschema','true')\
    .load('/home/aditya/github/Deep-learning-prerequisite/pyspark/Introduction/dataset/sf-fire-calls.csv')

spark_df.show()
````

    - Space in a column name is not a common thing.
    - We should fix the column names and remove the spaces from all the column names there are in the dataFrame.

### Transforming columns

Now this ```renamed_col_spark_df``` holds all the columns but the Call Number column just got renamed.

````python
renamed_col_spark_df.columns
````

### Chaining the withColumnRenamed to rename multiple columns in a spark's DataFrame
- You can create a chain of Spark transformation methods one after the other.
- Spark transformation returns a new dataFrame after transforming the old dataFrame.
- Spark dataFrame is immutable.
    - ```withColumnRenamed``` does not make changes in the original spark dataFrame but transforms the data and then create a new dataFrame.

### NOTE : 
spark DataFrame transformations will always produce a newly transformed dataFrame

````python
renamed_col_spark_df = spark_df \
                    .withColumnRenamed("CallNumber", "Call Number")\
                    .withColumnRenamed("UnitID","Unit ID")\
                    .withColumnRenamed("IncidentNumber","Incident Number")\
                    .withColumnRenamed("CallType","Call Type")\
                    .withColumnRenamed("CallDate","Call Date")\
                    .withColumnRenamed("WatchDate","Watch Date")\
                    .withColumnRenamed("CallFinalDisposition","Call Final Disposition")\
                    .withColumnRenamed("AvailableDtTm","Available Date Time")\
                    .withColumnRenamed("Address","Address")\
                    .withColumnRenamed("City","City")\
                    .withColumnRenamed("StationArea","Station Area")\
                    .withColumnRenamed("OriginalPriority","Original Priority")\
                    .withColumnRenamed("FinalPriority","Final Priority")\
                    .withColumnRenamed("ALSUnit","ALS Unit")\
                    .withColumnRenamed("CallTypeGroup","Call Type Group")\
                    .withColumnRenamed("NumAlarms","Num Alarms")\
                    .withColumnRenamed("UnitType","Unit Type")\
                    .withColumnRenamed("UnitSequenceInCallDispatch","Unit Sequence In CallDispatch")\
                    .withColumnRenamed("FirePreventionDistrict","Fire Prevention District")\
                    .withColumnRenamed("SupervisorDistrict","Supervisor District")\
                    .withColumnRenamed("RowID","Row ID")
````

- ```withColumn``` takes two arguments column_name and conversion_logic here in this case the   
    - column_name is ```CallDate```, ```WatchDate``` and ```AvailableDtTm``` 
    - transformation_logic is ```to_date("CallDate","MM/dd/yyyy")``` and ```to_date("AvailableDtTm","MM/dd/yyyy hh:mm;ss a")```

Summary : This code will change the dataFrame columns from string type to a date type and it will return a new dataFrame.

### Explaination : 
- ```cache()``` tells Spark to keep the DataFrame in memory (and sometimes on disk) after the first time it’s computed,
so future actions on the same DataFrame are much faster.
- Every time you run an action (like .show(), .count(), .collect()), Spark recomputes all the transformations from the beginning of the DAG (Directed Acyclic Graph).
- After the first action, Spark stores the resulting data in cluster memory (RAM).
Subsequent actions reuse that cached data instead of recomputing from the source.
    - Example : 
        - ```python
            spark_df.cache()
            spark_df.count()   # triggers computation + stores data in memory
            spark_df.show()    # now runs instantly from cache
          ```
- When you’re done with the DataFrame:
    - ```spark_df.unpersist()```

#### Q1. How many distinct types of calls were made to the fire department?
```sql
SELECT COUNT(DISTINCT CallType) AS Distinct_CallType_Count FROM fire_department_data_table
    WHERE callType IS NOT null;
```

**what do we want to do here?** <br>
- Filter the records and take only those where CallType is not null --> ```where()``` [TRANSFORMATION METHOD]
- Select the CallType column --> ```select()``` [TRANSFORMATION METHOD]
- Take only distinct call types --> ```distinct()``` [TRANSFORMATION METHOD]
- Show the count --> ```count()``` [ACTION METHOD]

````python
result_df = spark_df.where("CallType is not null")\
    .select("CallType")\
    .distinct()
print(f"distinct types of calls : {result_df.count()}")
````

### NOTE : 
- where, select and distinct are the transformation functions but count here is the action.
- An action doesn't return a dataFrame It returns the result to the driver.
- We cannot chain anything after we call an action method. Hence it is a best practice to keep the action methods on a separate line.

#### Q2. What are distinct types of calls made to the fire department?
```sql
SELECT DISTINCT CallType AS Distinct_CallTypes FROM fire_department_data_table
    WHERE CallType IS NOT null;
```

**Logic building**
- Filter the records based on CallType not null condition
- select the CallType column
- take only the distinct CallType

````python
result_df = spark_df.where("CallType is not null")\
    .select(expr("CallType AS distinct_call_type"))\
    .distinct()
result_df.show()
````

### Explaination : 
- ```expr("CallType AS distinct_call_type")``` In the select transformation method usually column name is given but here I wanted to rename the ```CallType``` column to ```distinct_call_type``` hence I had to use the expression ```expr```.
- I have to use the ```expr``` function to evaluate my expression.
- The ```select``` transformation method only takes a column name it doesn't take any expression if you want to use an expression in the select method you must have to evaluate the expression using ```expr``` function 

#### Q3. Find out all responses or delayed times greater than 5 mins?
```sql
SELECT CallNumber,Delay 
FROM fire_department_data_table WHERE Delay > 5 and CallType IS NOT null;
```

**Logic building:**
- Filter based on Delay where its greater than 5 seconds and CallType is not null
- select the columns "CallNumber" and "Delay"

````python
result_df = spark_df\
    .where("Delay > 5 and CallType IS NOT NULL")\
    .select("CallNumber","Delay")
result_df.show()
````

#### Q9. What week of the year in 2018 had the most fire calls?
```sql
    WITH most_fire_calls_2018_week AS (
        SELECT CallDate, CallType, COUNT(weekofyear(CallDate)) as call_count_by_week
        FROM fire_department_data_table 
        WHERE CallType LIKE '%Fire%' AND year(CallDate) = 2018
        GROUP BY CallDate,CallType
    )
    SELECT CallDate, CallType, call_count_by_week AS maximum_fire_call
    FROM most_fire_calls_2018_week ORDER BY call_count_by_week DESC 
    LIMIT 1;
```
