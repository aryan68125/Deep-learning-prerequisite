# Working with dataFrame in pySpark
## DataFrame Transformations and Actions (important for an interview perspective)
DataFrame offers a long list of methods however we can classify them into three main groups

- **Actions:** 
    - Actions trigger a spark job and return to the spark driver.
    - ![spark_execution_architecture](images/spark_execution_architecture.png)
    - Spark is a distributed processing system so it implements a driver and executor architecture in other words a master slave architecture.
    - So every spark program runs as a one driver and one or more executor.
    - Actions are dataFrame operations that **kick off a spark job** execution and return to the **Spark driver**.
- **Transformations:**
    - Spark dataFrame transformation produces a newly transformed dataFrame.
    - Transformations are different than Actions so they do not kick of a spark job and they do not return to the driver.
    - They simply create a new dataFrame and return it.
- **Functions/Methods:**
    - DataFrame methods or functions which are not categorized into Actions or transformations i.e everything which are not either Actions or Transformations are functions/methods

````python
from pyspark.sql import SparkSession

#  Create a spark session
spark = SparkSession.builder.appName("DfApp")\
        .config("spark.sql.ansi.enabled", "false")\
        .getOrCreate()

spark_df = spark.read.format('csv')\
    .option('header','true')\
    .option('inferschema','true')\
    .load('/home/aditya/github/Deep-learning-prerequisite/pyspark/Introduction/dataset/sf-fire-calls.csv')

spark_df.show()
````

### This dataFrame has two problems
- Column names are not standardized
    - Spark DataFrame's column names are case insensitive so upper case and lower case column names are the same.
    - Space in a column name is not a common thing.
    - We should fix the column names and remove the spaces from all the column names there are in the dataFrame.
    - Spark dataFrame provides you with column renamed transformation to rename the column.
- Date fields are of string type

````python
renamed_col_spark_df = spark_df \
                    .withColumnRenamed("Call Number", "CallNumber")
````

Now this ```renamed_col_spark_df``` holds all the columns but the Call Number column just got renamed.

### NOTICE: 
- In my findings I found out that pyspark 4.0.1 in my case automatically removed all the spaces from the column names which saved me from having to manually rename the column names manually in the dataFrame.

````python
renamed_col_spark_df.columns
````
