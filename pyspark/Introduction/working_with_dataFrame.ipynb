# Working with dataFrame in pySpark
## DataFrame Transformations and Actions (important for an interview perspective)
DataFrame offers a long list of methods however we can classify them into three main groups

- **Actions:** 
    - Actions trigger a spark job and return to the spark driver.
    - ![spark_execution_architecture](images/spark_execution_architecture.png)
    - Spark is a distributed processing system so it implements a driver and executor architecture in other words a master slave architecture.
    - So every spark program runs as a one driver and one or more executor.
    - Actions are dataFrame operations that **kick off a spark job** execution and return to the **Spark driver**.
- **Transformations:**
    - Spark dataFrame transformation produces a newly transformed dataFrame.
    - Transformations are different than Actions so they do not kick of a spark job and they do not return to the driver.
    - They simply create a new dataFrame and return it.
- **Functions/Methods:**
    - DataFrame methods or functions which are not categorized into Actions or transformations i.e everything which are not either Actions or Transformations are functions/methods

````python
from pyspark.sql import SparkSession

# imports related to spark dataFrame transformation
from pyspark.sql.functions import to_date, to_timestamp

#  Create a spark session
spark = SparkSession.builder.appName("DfApp")\
        .config("spark.sql.ansi.enabled", "false")\
        .getOrCreate()

spark_df = spark.read.format('csv')\
    .option('header','true')\
    .option('inferschema','true')\
    .load('/home/aditya/github/Deep-learning-prerequisite/pyspark/Introduction/dataset/sf-fire-calls.csv')

spark_df.show()
````

### This dataFrame has two problems
- Column names are not standardized
    - Spark DataFrame's column names are case insensitive so upper case and lower case column names are the same.
    - Space in a column name is not a common thing.
    - We should fix the column names and remove the spaces from all the column names there are in the dataFrame.
    - Spark dataFrame provides you with column renamed transformation to rename the column.
- Date fields are of string type

### NOTICE: 
- In my case the csv which I have column names that don't have spaces between them. hence I am going to do the opposite of what should be done to see if the transformation method actually works or not

````python
renamed_col_spark_df = spark_df \
                    .withColumnRenamed("CallNumber", "Call Number")
````

Now this ```renamed_col_spark_df``` holds all the columns but the Call Number column just got renamed.

````python
renamed_col_spark_df.columns
````

### Chaining the withColumnRenamed to rename multiple columns in a spark's DataFrame
- You can create a chain of Spark transformation methods one after the other.
- Spark transformation returns a new dataFrame after transforming the old dataFrame.
- Spark dataFrame is immutable.
    - ```withColumnRenamed``` does not make changes in the original spark dataFrame but transforms the data and then create a new dataFrame.

### NOTE : 
spark DataFrame transformations will always produce a newly transformed dataFrame

````python
renamed_col_spark_df = spark_df \
                    .withColumnRenamed("CallNumber", "Call Number")\
                    .withColumnRenamed("UnitID","Unit ID")\
                    .withColumnRenamed("IncidentNumber","Incident Number")\
                    .withColumnRenamed("CallType","Call Type")\
                    .withColumnRenamed("CallDate","Call Date")\
                    .withColumnRenamed("WatchDate","Watch Date")\
                    .withColumnRenamed("CallFinalDisposition","Call Final Disposition")\
                    .withColumnRenamed("AvailableDtTm","Available Date Time")\
                    .withColumnRenamed("Address","Address")\
                    .withColumnRenamed("City","City")\
                    .withColumnRenamed("StationArea","Station Area")\
                    .withColumnRenamed("OriginalPriority","Original Priority")\
                    .withColumnRenamed("FinalPriority","Final Priority")\
                    .withColumnRenamed("ALSUnit","ALS Unit")\
                    .withColumnRenamed("CallTypeGroup","Call Type Group")\
                    .withColumnRenamed("NumAlarms","Num Alarms")\
                    .withColumnRenamed("UnitType","Unit Type")\
                    .withColumnRenamed("UnitSequenceInCallDispatch","Unit Sequence In CallDispatch")\
                    .withColumnRenamed("FirePreventionDistrict","Fire Prevention District")\
                    .withColumnRenamed("SupervisorDistrict","Supervisor District")\
                    .withColumnRenamed("RowID","Row ID")
````

````python
renamed_col_spark_df.columns
````

````python
renamed_col_spark_df.show()
````

### Transforming date fields from string type to date type

````python
spark_df.printSchema()
````

````python
spark_df = spark_df\
    .withColumn("CallDate",to_date("CallDate","MM/dd/yyyy"))\
    .withColumn("WatchDate", to_date("WatchDate","MM/dd/yyyy"))\
    .withColumn("AvailableDtTm", to_date("AvailableDtTm","MM/dd/yyyy hh:mm;ss a"))
````
